```python
import warnings
warnings.filterwarnings("ignore")

# Logistic Regression (classification)

# We use L2 penalty (default) because it's stable for text.
logreg_l2 = LogisticRegression(
    penalty="l2",
    C=1,          # inverse of regularization strength (smaller C => stronger regularization)
    solver="liblinear",
    max_iter=2000
)

logreg_l2.fit(X_train, y_train)

# Get predicted class labels (strings) for confusion matrix and classification report
predicted_labels = logreg_l2.predict(X_test)

# Get probabilities for all classes
all_class_probabilities = logreg_l2.predict_proba(X_test)

# Determine the mapping from original labels to binary (0/1) for ROC-AUC and Log Loss
# The model's class order is crucial here
model_classes = logreg_l2.classes_ # e.g., array(['c', 'j'], dtype='<U1')

# Map y_test to binary (0 for first class, 1 for second class in model_classes)
# This assumes model_classes has two elements and that the second element corresponds to the 'positive' class
# for which predict_proba returns the second column (p_test).
y_test_binary = np.array([1 if label == model_classes[1] else 0 for label in y_test])

# Extract probabilities for the 'positive' class (class corresponding to index 1)
p_test = all_class_probabilities[:, 1]

print("Probabilities (first 8):", np.round(p_test[:8], 3))
print("Predictions   (first 8):", predicted_labels[:8]) # Use string predictions
print("True labels   (first 8):", y_test[:8])

print("\nConfusion matrix (LogReg L2, threshold=0.5):")
# UserWarning: A single label was found in 'y_true' and 'y_pred'.
# This warning occurs because the test set contains only one class.
print(confusion_matrix(y_test, predicted_labels)) # Use string labels

print("\nClassification report:")
print(classification_report(y_test, predicted_labels, digits=3)) # Use string labels

# Check if y_test_binary has at least two unique labels for ROC-AUC and Log Loss
if len(np.unique(y_test_binary)) < 2:
    print("\nROC-AUC: Not applicable (test set contains only one class).")
    print("Log loss: Not applicable (test set contains only one class).")
    auc = np.nan
    ll = np.nan
else:
    # ROC-AUC evaluates ranking quality (higher is better, 0.5 is random)
    auc = roc_auc_score(y_test_binary, p_test) # Use binary y_test for this metric
    print("ROC-AUC:", round(auc, 3))

    # Log loss punishes confident wrong predictions heavily
    ll = log_loss(y_test_binary, p_test) # Use binary y_test for this metric
    print("Log loss:", round(ll, 3))
```
